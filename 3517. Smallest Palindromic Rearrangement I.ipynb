```python
def smallestPalindrome(s):
    n = len(s)
    res = ''
    if n % 2 == 0:
        res = "".join(sorted(s[:n//2]))
        return res + res[::-1]
    else:
        res = "".join(sorted(s[:n//2]))
        return res + s[n//2] + res[::-1]
```

```python
s = "daccad"
smallestPalindrome(s)
```

```python
s = "z"
smallestPalindrome(s)
```
