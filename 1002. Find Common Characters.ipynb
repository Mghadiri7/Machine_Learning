```python
from collections import Counter

def commonChars(words):
    shortest = min(words, key=len)
    common = Counter(shortest)
    for word in words:
        common &= Counter(word)
    return list(common.elements())
```

```python
words = ["bella","label","roller"]
commonChars(words)
```

```python
words = ["cool","lock","cook"]
commonChars(words)
```
