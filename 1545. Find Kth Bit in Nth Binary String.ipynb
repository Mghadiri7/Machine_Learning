## Frist Solution

```python
def findKthBit(n, k):
    def kth(x):
        if x == 1:
            return "0"
        else:
            _str = kth(x-1)
            inverted_str = "".join("1" if bit == "0" else "0" for bit in _str)
            return _str + "1" + inverted_str[::-1]

    return kth(n)[k-1]
```

```python
n = 4
k = 11
findKthBit(n, k)
```

## Second Solution

```python
def findKthBit(n, k):
    if n == 1:
        return '0'
    midindex = pow(2, n-1)
    if k == midindex:
        return '1'
    elif k < midindex:
        return findKthBit(n-1, k)
    else:
        return '0' if findKthBit(n-1, midindex * 2 - k) == '1' else '1'
```

```python
n = 4
k = 11
findKthBit(n, k)
```
